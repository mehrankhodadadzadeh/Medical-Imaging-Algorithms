# 1. Image Enhancement
## 1.1 The image histogram
The histogram represents how many pixels have a certain intensity in the corresponding image. In image processing, it facilitates the identification of image acquisition issues, for example:

- **Over and under exposure:** Are intensity values spread out (good) or clustered (bad)?

<img src="./images/hist_exposure.png" alt="drawing" width="800"/>

- **Contrast:** In the image, are there many distinct intensity values (high contrast), or does the image use few intensity values (low contrast)? A "normal" contrast is when intensity values are widely spread, with a significant difference between min and max intensity values.

<img src="./images/hist_contrast.png" alt="drawing" width="800"/>

- **Dynamic range:** Related to the number of distinct pixels in the image.

<img src="./images/hist_dyn_range.png" alt="drawing" width="800"/>

Unlike previous examples, medical images can have a large intensity range or even floating point intensities. This yields very large histograms and makes the pixel count per intensity impractical.

<img src="./images/hist_mri.png" alt="drawing" width="600"/>

Therefore, in practice intensities are usually binned, i.e. grouped in a reduced number of bins with similar intensity.

## 1.2 Image enhancement

We shall discuss three ways of contrast improvement:

1. [Linear contrast mapping](http://homepages.inf.ed.ac.uk/rbf/HIPR2/stretch.htm) or histogram stretching. It involves a linear transformation on the image intensities, such that the transformed intensities cover to the full range.
2. [Histogram equalisation](https://homepages.inf.ed.ac.uk/rbf/HIPR2/histeq.htm). In this case, the aim is to obtain a uniform histogram, in which all intensities are equally represented. This can be done by applying a nonlinear transformation on the image intensities.
3. [CLAHE](https://docs.opencv.org/4.x/d5/daf/tutorial_py_histogram_equalization.html). Contrast Limiting Adaptive Histogram Enhancement (CLAHE) solves the resultant noise from applying histogram equalization. Instead of executing the equalization in a "global" manner, CLAHE applies histogram equalization in small patches taken from the image.

````python
#Exercise 1.1

# Function(linear contrast stretching)
def linear_contrast_stretching(image_array):
    #the 5th and 95th percentiles
    p5, p95 = np.percentile(image_array, (5, 95))
    image_clipped = np.clip(image_array, p5, p95)

    # Apply it
    image_stretched = (image_clipped - p5) / (p95 - p5)

    return image_stretched

# And apply that to the image
stretched_image = linear_contrast_stretching(image_array)


#Exercise 1.2:Histogram equalization

# perform histogram equalization(function)
def histogram_equalization(image_array):
    hist, bin_edges = np.histogram(image_array.ravel(), bins=64)
    # Calculate the center
    bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2

    #the cumulative distribution function
    cdf = hist.cumsum()
    # Normalize the cdf by dividing to the maximum to scael it between 0 and 1
    cdf_normalized = cdf / cdf[-1]

    #map the original pixel values to new values
    image_eq = np.interp(image_array.ravel(), bin_centers, cdf_normalized)

    # Reshape to the original image shape
    image_eq_reshaped = image_eq.reshape(image_array.shape)

    return image_eq_reshaped

# Applying histogram equalization to image
image_eq_reshaped = histogram_equalization(image_array)


#Report


# Apply CLAHE (Contrast Limited Adaptive Histogram Equalization)
#Adjusting the `clip_limit` in CLAHE can change contrast enhancement;
# decreasing it conserves contrast and minimizes noise while,
# increasing it amplifies contrast and features but risks enhancing noise.
clahe = exposure.equalize_adapthist(image_array,clip_limit=0.04)

#subplots
fig, axs = plt.subplots(4, 2, figsize=(12, 20))

#Original Image and Histogram
axs[0, 0].imshow(image_array, cmap='gray')
axs[0, 0].set_title('Original Image')
axs[0, 1].hist(image_array.ravel(), bins='auto')
axs[0, 1].set_title('Histogram of Original Image')

# Linear Stretch and H
axs[1, 0].imshow(stretched_image, cmap='gray')
axs[1, 0].set_title('Linear Stretch Image')
axs[1, 1].hist(stretched_image.ravel(), bins='auto')
axs[1, 1].set_title('Histogram of Linear Stretch Image')

# Histogram Equalized Image and H
axs[2, 0].imshow(image_eq_reshaped, cmap='gray')
axs[2, 0].set_title('Histogram Equalized Image')
axs[2, 1].hist(image_eq_reshaped.ravel(), bins=64)
axs[2, 1].set_title('Histogram of Histogram Equalized Image')

# CLAHE Image and H
axs[3, 0].imshow(clahe, cmap='gray')
axs[3, 0].set_title('CLAHE Image')
axs[3, 1].hist(clahe.ravel(), bins='auto')
axs[3, 1].set_title('Histogram of CLAHE Image')

# Adjust layout
plt.tight_layout()
plt.show()
````

# 2. Image Denoising

Acquiring an image is always prone to artifacts that may corrupt or degrade its quality. Examples of them are noise, blurring, and distortion. Multiple image restoration filters have been proposed in medical imaging to reduce the effect of these artifacts. Image filters may be used either to improve the image quality before reviewing it or as a pre-processing step to improve the result of further image processing operations such as segmentation.

<img src="./images/denoising.png" alt="drawing" width="500"/>

## 2.1 Noise suppression.

Image noise can often be assumed to be a high-frequency signal. Therefore, many noise reduction approaches filter the high-frequency components while preserving the low-frequency ones; a typical example is the 2D-Gaussian filter.

Despite the wide use of low-pass filtering, this technique has the side effect of blurring the edges of the image. To avoid it, smoothing filters that preserve the edges, such as the non-linear median filter, have been proposed.

<img src="./images/noise_removal.png" alt="drawing" width="700"/>

## 2.2 Edge enhancement
The goal is to enhance the edge contrast of an image in an attempt to improve its apparent sharpness. The resultant edge-image can be added to the original image to enhance the visual quality or be employed as input in an image segmentation approach.

<img src="./images/edge_enhancement.png" alt="drawing" width="700"/>

````python
from scipy import ndimage
from skimage.filters import prewitt

#Function to find RMSE
def rmse(imageA, imageB):
    diff = imageA - imageB
    sq_diff = np.square(diff)
    mse = np.mean(sq_diff)
    return np.sqrt(mse)

#apply filters and calculating metrics
def apply_filter_and_metrics(noisy_image, ground_truth, filter_type, **kwargs):
    if filter_type == 'gaussian':
        filtered_image = ndimage.gaussian_filter(noisy_image, **kwargs)
    elif filter_type == 'median':
        filtered_image = ndimage.median_filter(noisy_image, **kwargs)
    elif filter_type == 'mean':
        kernel = np.ones(kwargs['size']) / np.prod(kwargs['size'])
        filtered_image = ndimage.convolve(noisy_image, kernel)

    # Calculating a filtered noise and remaining noise
    filtered_noise = noisy_image - filtered_image
    remaining_noise = ground_truth - filtered_image

    # Calculate RMSE
    error = rmse(filtered_image, ground_truth)

    # Create edge map with using Prewitt function
    edge_map = prewitt(filtered_image)

    return filtered_image, filtered_noise, remaining_noise, edge_map, error

# Reading Images
ground_truth = sitk.GetArrayFromImage(sitk.ReadImage('/content/T2.mha'))
noisy_image = sitk.GetArrayFromImage(sitk.ReadImage('/content/SP.mha'))

# Apply filters
results = {}
results['gaussian'] = apply_filter_and_metrics(noisy_image, ground_truth, 'gaussian', sigma=1)
results['median'] = apply_filter_and_metrics(noisy_image, ground_truth, 'median', size=3)
results['mean'] = apply_filter_and_metrics(noisy_image, ground_truth, 'mean', size=(3, 3))

# Display
fig, axs = plt.subplots(3, 4, figsize=(20, 15))
filter_types = ['gaussian', 'median', 'mean']
for i, filter_type in enumerate(filter_types):
    axs[i, 0].imshow(results[filter_type][0], cmap='gray')
    axs[i, 0].set_title(f'{filter_type.capitalize()} Filtered Image')
    axs[i, 1].imshow(results[filter_type][1], cmap='gray')
    axs[i, 1].set_title(f'{filter_type.capitalize()} Filtered Noise')
    axs[i, 2].imshow(results[filter_type][2], cmap='gray')
    axs[i, 2].set_title(f'{filter_type.capitalize()} Remaining Noise')
    axs[i, 3].imshow(results[filter_type][3], cmap='gray')
    axs[i, 3].set_title(f'{filter_type.capitalize()} Edge Map')
    print(f'RMSE for {filter_type} filter:', results[filter_type][4])

plt.tight_layout()
plt.show()



````

## 2.3 Intensity non-uniformity correction

Non-uniform intensity correction is another common task in image denoising. Grayscale inhomogeneities appear in magnetic resonance (MR) images as systematic changes in the local statistical characteristics of tissues. Homomorphic Unsharp Masking (HUM) is applied as a post-processing tool to reduce these intensity effects.

HUM is conceptually straightforward; it can be easily implemented and is very fast. It relies on the assumption that if grayscale inhomogeneities are not present in the image, the mean or median in a local window should match the global mean or median of the overall image. This assumption is approximately true when the filter window is large enough to enclose a representative sample of tissues.

For a detailed implementation see the paper: [*''Optimized Homomorphic Unsharp Masking for MR Grayscale Inhomogeneity Correction'' by Benjamin H. Brinkmann, Armando Manduca and Richard A. Robb, IEEE, 1998*](https://ieeexplore.ieee.org/document/700729)

HUM requires the computation of:
- The global mean value $\mu$ of the corrupted image
- The local mean values $\mu_{i,j}$ for each pixel considering a neighbourhood
- The HUM corrected/ideal value of a pixel is:

$$f_{i,j} = g_{i,j} \cdot \frac{\mu}{\mu_{i,j}},$$

   where $g_{i,j}$ is the intensity value of the input image (corrupted/observed image).

````python
from skimage.metrics import structural_similarity as ssim
from scipy.ndimage import uniform_filter

def calculate_nrmse(imageA, imageB):
    diff = imageA.astype(np.float32) - imageB.astype(np.float32)
    norm_diff = diff / np.max(imageB)
    mse = np.mean(np.square(norm_diff))
    return np.sqrt(mse)

def hum_algorithm(image, window_size, pad_size=0, global_threshold=10):  # Set default threshold to 10
    # Padding, if specified
    if pad_size > 0:
        image = np.pad(image, pad_size, mode='constant')

    # finding global mean with threshold
    valid_pixels = image > global_threshold  # Apply threshold
    global_mean = np.mean(image[valid_pixels])

    local_mean = uniform_filter(image, size=window_size)
    epsilon = 1e-6
    corrected_image = image * global_mean / (local_mean + epsilon)

    if pad_size > 0:
        corrected_image = corrected_image[pad_size:-pad_size, pad_size:-pad_size]

    return np.clip(corrected_image, 0, 255).astype(np.uint8)

# Reading images
non_bias_image = sitk.GetArrayFromImage(sitk.ReadImage('/content/Flair.mha', sitk.sitkUInt8))
bias_image = sitk.GetArrayFromImage(sitk.ReadImage('/content/Bias.mha', sitk.sitkUInt8))

# Apply HUM in different ways
window_size = 40
corrected_basic = hum_algorithm(bias_image, window_size)
corrected_padded = hum_algorithm(bias_image, window_size, pad_size=window_size//2)
corrected_threshold = hum_algorithm(bias_image, window_size, pad_size=window_size//2, global_threshold=10)

# Evaluate a performance
nrmse_basic = calculate_nrmse(non_bias_image, corrected_basic)
nrmse_padded = calculate_nrmse(non_bias_image, corrected_padded)
nrmse_threshold = calculate_nrmse(non_bias_image, corrected_threshold)

ssim_basic = ssim(non_bias_image, corrected_basic)
ssim_padded = ssim(non_bias_image, corrected_padded)
ssim_threshold = ssim(non_bias_image, corrected_threshold)

# Plotting results
fig, axs = plt.subplots(1, 4, figsize=(20, 5))
axs[0].imshow(bias_image, cmap='gray')
axs[0].set_title('Biased Image')
axs[1].imshow(corrected_basic, cmap='gray')
axs[1].set_title(f'Basic HUM\nNRMSE: {nrmse_basic:.3f}\nSSIM: {ssim_basic:.3f}')
axs[2].imshow(corrected_padded, cmap='gray')
axs[2].set_title(f'Padded HUM\nNRMSE: {nrmse_padded:.3f}\nSSIM: {ssim_padded:.3f}')
axs[3].imshow(corrected_threshold, cmap='gray')
axs[3].set_title(f'Threshold HUM\nNRMSE: {nrmse_threshold:.3f}\nSSIM: {ssim_threshold:.3f}')

for ax in axs:
    ax.axis('off')
plt.tight_layout()
plt.show()
````
